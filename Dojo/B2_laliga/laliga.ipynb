```python
 # B2 · La Liga Table

**Qué resuelve:** partiendo de un diccionario de resultados de partidos de La Liga, calcular la tabla de clasificación: puntos, partidos jugados, victorias, empates y derrotas por equipo. Ordenar descendente por puntos y mostrar formateado.

**Qué aprendí:**
- Nested dicts: `stats[team]['points']`
- Patrón **setup antes de operación** (inicializar entrada del dict antes de `+=`)
- Formateo de output con f-strings (`{team:<15} | {points:<4}`)
- `sorted` con `lambda item: item[1]['points']` sobre dict anidado
- Separar función de cálculo de función de impresión (semilla MVC)

**Bugs que me pillaron:**
Ver `/CUIDADO_RESBALADIZO.md` en la raíz:
- Setup antes de operación (dicts) → KeyError si no inicializas
- Olvidar llamar la función de impresión al final del script

**Fecha:** octubre-noviembre 2026 · Curso IBM DS · Course 4

#tabla de La Liga (españa)

resultados = {
    ('Real Madrid', 'Barcelona'): (1, 2),
    ('Atlético', 'Athletic'): (1, 1),
    ('Barcelona', 'Atlético'): (3, 0),
    ('Athletic', 'Real Madrid'): (0, 2),
    ('Real Madrid', 'Atlético'): (1, 1)
}

def calculate_points(results):
    points = {}
    for (team1, team2), (score1, score2) in results.items():
        # Inicializar siempre, no solo en empate
        if team1 not in points:
            points[team1] = 0
        if team2 not in points:
            points[team2] = 0

        if score1 > score2:
            points[team1] += 3
        elif score1 < score2:
            points[team2] += 3
        else:
            points[team1] += 1
            points[team2] += 1

    return points

    return points

    
#returns played_games, wins, draws, losses, points for each team in a dictionary
def calculate_team_stats(results):
    stats = {}
    for (team1, team2), (score1, score2) in results.items():
        if team1 not in stats:
            stats[team1] = {'played': 0, 'wins': 0, 'draws': 0, 'losses': 0, 'points': 0}
        if team2 not in stats:
            stats[team2] = {'played': 0, 'wins': 0, 'draws': 0, 'losses': 0, 'points': 0}

        stats[team1]['played'] += 1
        stats[team2]['played'] += 1

        if score1 > score2:
            stats[team1]['wins'] += 1
            stats[team2]['losses'] += 1
            stats[team1]['points'] += 3
        elif score1 < score2:
            stats[team2]['wins'] += 1
            stats[team1]['losses'] += 1
            stats[team2]['points'] += 3
        else:
            stats[team1]['draws'] += 1
            stats[team2]['draws'] += 1
            stats[team1]['points'] += 1
            stats[team2]['points'] += 1

    return stats

#ejecutar!


#Bulding the league table
league_table = calculate_team_stats(resultados)

#Diccionario ordenado de forma descendente por puntos usando sorted y lambda
sorted_league_table = dict(sorted(
    league_table.items(), key=lambda item: item[1]['points'], 
    reverse=True
    ))
#fucion para mostrar la tabla de La Liga
def print_sorted_table(table):
    print("--- CLASIFICACIÓN DE LA LIGA ---")
    print(f"{'Equipo':<15} | {'Pts':<4} | {'PJ':<4} | {'G':<4} | {'E':<4} | {'P':<4}")
    print("-" * 45)
    for team, stats in table.items():
        print(f"{team:<15} | {stats['points']:<4} | {stats['played']:<4} | {stats['wins']:<4} | {stats['draws']:<4} | {stats['losses']:<4}")    

#tabla de La Liga (españa)
#juegos jugados, puntos, victorias, empates, derrotas sorted labda mayor a menor 
print_sorted_table(sorted_league_table)

```
