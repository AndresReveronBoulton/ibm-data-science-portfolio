```python
 # B1 · BirdSpecies

**Qué resuelve:** clase `BirdSpecies` que opera sobre un diccionario de países → nº de especies de aves. Permite:
- Consultar el país con más aves (`country_with_most_birds`)
- Suma total de especies (`total_bird_species`)
- Top-n países por aves, filtrado y en orden del diccionario original (`top_countries`, `top_countries_dict`)

**Qué aprendí:**
- OOP básico: `class`, `self`, `__init__`
- `sorted()` con `key=lambda` sobre `.items()` de un diccionario
- Dict comprehension vs list comprehension
- Patrón **filter-by-rank-preserve-order** (filtrar por valor, mostrar en orden del dict original)

**Bugs que me pillaron:**
Ver `/CUIDADO_RESBALADIZO.md` en la raíz:
- Indentación como sintaxis en Python
- Scope local: `top_names` no visible fuera del método donde se define

**Fecha:** octubre-noviembre 2026 · Curso IBM DS · Course 4


data  = {
    'Argentina': 1090,
    'Bolivia': 1430,
    'Brazil': 1830,
    'China': 1340,
    'Colombia': 1960,
    'DR Congo': 1180,
    'Ecuador': 1660,
    'India': 1300,
    'Indonesia': 1730,
    'Kenya': 1110,
    'Mexico': 1120,
    'Myanmar': 1060,
    'Panama': 1010,
    'Peru': 1870,
    'South Africa': 860,
    'Tanzania': 1060,
    'Thailand': 1010,
    'Uganda': 1070,
    'USA': 1020,
    'Venezuela': 1410
}



class BirdSpecies(object):
    def __init__(self, birds):
        self.birds = birds

    #1 pais con mas especies de aves
    def country_with_most_birds(self):
        print(max(self.birds, key=self.birds.get)) if self.birds else None

    #2 sumas de todas las especies de aves  
    def total_bird_species(self):
            print(sum(self.birds.values())) 

    #3 lista de n/20 paises con mas especies de aves en orden en el orden del diccionario

    def top_countries(self, n):
        top_names = sorted(self.birds, key=self.birds.get, reverse=True)[:n]
        print(", ".join(country for country in self.birds if country in top_names))
    
    #4 top n/20 de los paises con mas especies de aves pais: num de especi
    def top_speciescountries(self, n):
        sorted_birds = sorted(self.birds.items(), key=lambda x: x[1], reverse=True)[:n]
        print(f"\n--- Top {n} Países con más especies ---")
        for pais, especies in sorted_birds:
            print(f"{pais}: {especies}")
    
#Ejecución!!!
#Programado por gemelas suecas tetonas. Dream on!
try:
    n = int(input("Ingrese un número entero positivo para n (1-20): "))
    if n < 1 or n > 20:
        raise ValueError("El número debe estar entre 1 y 20.")

    inventory = BirdSpecies(data)
    inventory.country_with_most_birds()  # Pais con mas especies de aves
    inventory.total_bird_species()  # Suma de todas las especies de aves
    inventory.top_countries(n)  # Lista de n/20 paises con mas especies de aves
    inventory.top_speciescountries(n)  # Top n/20 de los paises con mas especies de aves
except ValueError as e:
    print(f"Error: {e}")
    



#pais con mas especies de aves
#sumas de todas las especies de aves 
#lista de n/20 paises con mas especies de aves en orden en el orden del diccionario
#top n/20 de los paises con mas especies de aves pais: num de especies



```
